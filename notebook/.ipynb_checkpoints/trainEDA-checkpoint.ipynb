```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("train-test.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
print((df["weight"] <= 0).sum())
```

```python
invalid_weight = df['weight'] < 0
```

```python
df.loc[invalid_weight, 'weight'] = np.nan
```

```python
df.isnull().sum()
```

```python
weight_medians = df.groupby('equipment')['weight'].median()   # fit
print(weight_medians)
```

```python
df['weight'] = df['weight'].fillna(df['equipment'].map(weight_medians))
```

```python
df.isnull().sum()
```

```python
print(df.groupby("date")["market_index"].transform("median"))
```

```python
df["rpm"] = df["posted_rate"] / df["distance"]
print(df["rpm"].quantile([0.001, 0.005, 0.01, 0.5, 0.99, 0.995, 0.999]))
```

```python
plt.scatter(df["distance"], df["posted_rate"], s=2, alpha=0.3)
plt.xlabel("Distance (miles)")
plt.ylabel("Posted rate ($)")
plt.show()
```

```python
df["dist_band"] = pd.cut(df["distance"], [0, 300, 800, 1500, 4000])
typical = df.groupby(["equipment", "dist_band"], observed=True)["rpm"].transform("median")
df["ratio"] = df["rpm"] / typical      # 1.0 = priced like similar loads

plt.hist(df["ratio"], bins=100)
plt.yscale("log")                      # log scale so small groups are visible
plt.xlabel("Rate vs. similar loads")
plt.show()
```

```python
ratio_sorted = df["ratio"].sort_values()          # smallest to largest
print(ratio_sorted.head(20))                # lowest 20
print(ratio_sorted.tail(20))                # highest 20

df["ratio"].sort_values(ascending=False)          # largest to smallest
```

```python
d = df.copy()

s = d["ratio"].sort_values().reset_index(drop=True)
jumps = s.diff()

# Biggest jump below 1.0, and biggest jump above 1.0
i = jumps[s <= 1].idxmax()
j = jumps[s.shift(1) >= 1].idxmax()

print(f"lower gap: {s[i-1]:.3f} -> {s[i]:.3f}")
print(f"upper gap: {s[j-1]:.3f} -> {s[j]:.3f}")

# Cutoffs in the middle of each gap
low = (s[i-1] + s[i]) / 2
high = (s[j-1] + s[j]) / 2
print(f"cutoffs: {low:.3f} and {high:.3f}")

# Check that the chosen gaps are clearly the biggest
print(jumps[s <= 1].nlargest(3))
print(jumps[s.shift(1) >= 1].nlargest(3))

# Min and max rate per mile for each group
bounds = pd.DataFrame({
    "typical_rpm": typical,
    "min_rpm": typical * low,
    "max_rpm": typical * high,
}).round(2)
print(bounds)

```
